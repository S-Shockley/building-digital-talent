```python
#Load the dataset

df_2025 = pd.read_csv("../data/raw/results2025.csv")

df_2025.shape
```

The 2025 Stack Overflow Developer Survey contains 49,191 responses and 172 columns. The full dataset includes demographic information, education, learning pathways, employment characteristics, compensation, AI attitudes, job satisfaction, and career change consideration.

```python
project_cols = [
    "ResponseId",
    "JobSat",
    "NewRole",
    "LearnCodeChoose",
    "LearnCode",
    "LearnCodeAI",
    "EdLevel",
    "YearsCode",
    "WorkExp",
    "ConvertedCompYearly",
    "Age",
    "Employment",
    "DevType",
    "OrgSize",
    "RemoteWork",
    "Country",
    "AISelect",
    "AIThreat"
]

df_projresult25 = df_2025[project_cols].copy()

df_projresult25.head()
```

```python
df_projresult25.shape
```

The missing value counts below indicate that missing values are significant and a strategy needs to be developed to handle them

```python
missing = (
    df_projresult25.isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame("missing_count")
)

missing["missing_percent"] = (
    missing["missing_count"] / len(df_projresult25) * 100
).round(1)

missing
```

the next cell creates the target variable of high job satisfaction where job sat of greater than 8 is 1. Anything lower is 0.

```python
df_projresult25["high_job_sat"] = np.where(
    df_projresult25["JobSat"] >= 8,
    1,
    0
)

df_projresult25.loc[df_projresult25["JobSat"].isna(), "high_job_sat"] = np.nan
```

The next cell creates the target variable for considering a career change

```python
df_projresult25["career_change_considered"] = np.where(
    df_projresult25["NewRole"].str.contains("neither", case=False, na=False),
    0,
    1
)

df_projresult25.loc[df_projresult25["NewRole"].isna(), "career_change_considered"] = np.nan

df_projresult25["career_change_considered"].value_counts(dropna=False)
```

If the answer contains the word "neither", code it as 0 because they did not consider or transition. Otherwise, code it as 1 because they either considered a change or already transitioned. Then put missing answers back to missing.

If somebody did not answer the job satisfaction question, they cannot be considered for the job satisfaction model.

```python
df_jobsat_model = df_projresult25.dropna(subset=["JobSat"]).copy()
```

If somebody did not answer the career change question, they cannot be considered for the classification model

```python
df_newrole_model = df_projresult25.dropna(subset=["NewRole"]).copy()
```

```python
categorical_cols = [
    "LearnCodeChoose",
    "LearnCode",
    "LearnCodeAI",
    "EdLevel",
    "Age",
    "Employment",
    "DevType",
    "OrgSize",
    "RemoteWork",
    "Country",
    "AISelect",
    "AIThreat"
]
```

```python
df_projresult25["active_learner"] = np.where(
    df_projresult25["LearnCodeChoose"].str.contains(
        "not new to coding but am learning",
        case=False,
        na=False
    ),
    1,
    0
)

df_projresult25.loc[
    df_projresult25["LearnCodeChoose"].isna()
    | df_projresult25["LearnCodeChoose"].eq("Missing"),
    "active_learner"
] = np.nan

df_projresult25["active_learner"].value_counts(dropna=False)
```

Missingness may actually be meaningful.
It avoids dropping thousands of rows.
One-hot encoding can handle "Missing" later.

```python
for col in categorical_cols:
    df_projresult25[col] = df_projresult25[col].fillna("Missing")
```

Because salary and experience can have extreme outliers. Median is safer.

```python
df_projresult25["log_salary"] = np.log1p(df_projresult25["ConvertedCompYearly"])
```

the cell above with log_salary is a transformed version of salary. It shrinks very large salary values so extreme outliers do not overpower the model. 

Compensation was transformed using a natural log transformation to reduce the influence of extreme salary outliers while preserving the relative ordering of salaries.

```python
df_projresult25["WorkExp_clean"] = pd.to_numeric(
    df_projresult25["WorkExp"],
    errors="coerce"
)
```

```python
numeric_cols = [
    "YearsCode_clean",
    "WorkExp_clean",
    "log_salary"
]
```

```python
def clean_years_code(value):
    if pd.isna(value):
        return np.nan
    if value == "Less than 1 year":
        return 0.5
    if value == "More than 50 years":
        return 51
    return float(value)

df_projresult25["YearsCode_clean"] = df_projresult25["YearsCode"].apply(clean_years_code)
```

```python
for col in numeric_cols:
    df_projresult25[col] = df_projresult25[col].fillna(df_projresult25[col].median())
```
